**File:** 2.1_text_data.txt  
**Commit Date:** 11/aug/2026

```python
import string
import re

# Input text for Exp 2.1
raw_text = """Natural Language Processing (NLP) is a branch of Artificial Intelligence.  It helps computers understand and process human language.

In 2026, NLP is used in many applications such as chatbots, machine translation, sentiment analysis, and text summarization!!!

Text preprocessing is an important step in NLP.   It includes several operations such as cleaning text, removing punctuation, and removing unnecessary numbers.

For example, a chatbot may process 1000+ customer messages every day. These messages can contain CAPITAL letters, numbers like 12345, punctuation marks!!!, and   extra spaces.

The main goal of text preprocessing is to convert raw text into a clean and consistent format. This makes the text easier to analyze and use for further Natural Language Processing tasks."""

def perform_preprocessing(text):
    # 1. Lowercase and counting Uppercase
    upper_count = sum(1 for c in text if c.isupper())
    text_lower = text.lower()

    # 2. Punctuation removal and counting types
    punctuations_found = [c for c in text if c in string.punctuation]
    punc_types = len(set(punctuations_found))
    text_no_punc = text_lower.translate(str.maketrans('', '', string.punctuation))

    # 3. Number removal and counting numbers
    numbers = re.findall(r'\d', text_no_punc)
    text_no_nums = re.sub(r'\d', '', text_no_punc)

    # 4. Extra Whitespace removal and counting
    extra_spaces = len(re.findall(r'\s{2,}', text_no_nums))
    text_cleaned = " ".join(text_no_nums.split())

    print(f"Uppercase letters: {upper_count}")
    print(f"Different types of punctuation: {punc_types}")
    print(f"Count of numbers: {len(numbers)}")
    print(f"Extra whitespace present: {extra_spaces}")
    print(f"\n--- Cleaned Text ---\n{text_cleaned}")
    return text_cleaned

cleaned_data = perform_preprocessing(raw_text)
```

### Note on Experiment 2.1: Basic Preprocessing
- **Lowercase**: Converts all characters to the same case to ensure the model treats 'Apple' and 'apple' as the same word.
- **Punctuation**: Removed using `string.punctuation` because symbols usually don't carry semantic meaning for basic tasks.
- **Numbers**: Stripped using Regular Expressions (`\d`).
- **Whitespace**: Handled by splitting the string and joining with a single space to remove redundant gaps.

```python
import nltk
import spacy
import re

# Download resources
nltk.download('punkt')
nltk.download('punkt_tab')
nlp = spacy.load("en_core_web_sm")

text = "Natural Language Processing (NLP) is a branch of Artificial Intelligence. In 2026, NLP is used!"

# Method 1: Without NLTK/spaCy
words_manual = text.split()
sentences_manual = re.split(r'(?<=[.!?]) +', text)

# Method 2: NLTK
from nltk.tokenize import word_tokenize, sent_tokenize
words_nltk = word_tokenize(text)
sentences_nltk = sent_tokenize(text)

# Method 3: spaCy
doc = nlp(text)
words_spacy = [t.text for t in doc]
sentences_spacy = [s.text for s in doc.sents]

print("Manual Words:", words_manual)
print("NLTK Sentences:", sentences_nltk)
print("spaCy Words:", words_spacy)
```

### Experiment 2.2: Tokenization
**File:** 2.2_tokenization_data.txt

- **Manual**: splitting by whitespace.
- **NLTK**: standard linguistic tokenization.
- **spaCy**: industrial-grade tokenization.

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import nltk
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))
input_text = "Natural Language Processing (NLP) is a fascinating field of Artificial Intelligence. It enables computers to understand human language. NLP is used in many real-world applications, such as chatbots, machine translation, sentiment analysis, and text summarization.\n\nText tokenization is one of the first steps in NLP. Sentence tokenization divides a paragraph into individual sentences. Word tokenization divides each sentence into individual words and tokens.\n\nFor example, a chatbot receives thousands of messages every day. The system processes these messages and identifies useful information. This information can then be used to provide better responses to users."
tokens = word_tokenize(input_text.lower())

# Filter tokens
filtered_tokens = [w for w in tokens if w.isalpha() and w not in stop_words]
removed_stops = [w for w in tokens if w in stop_words]

print(f"Original Tokens: {tokens}")
print(f"Stop Words Removed: {removed_stops}")
print(f"Filtered Tokens (Clean): {filtered_tokens}")
```
